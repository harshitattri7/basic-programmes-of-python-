```python
labels=["products"]
online=[60]
offline=[40]
free=[50]
plt.bar(labels,online,label="online")
plt.bar(labels,offline,bottom=online,label="offline")
plt.bar(labels,free,bottom=[online[0]+offline[0]],label="free")
plt.title('stacked bar graph')
plt.legend()
plt.show()
```

```python
labels=["samsung","apple","others"]
online=[60,50,40]
offline=[75,80,40]
free=[50,40,80]
plt.bar(labels,online,label="online")
plt.bar(labels,offline,bottom=online,label="offline")
plt.bar(labels,free,bottom=online + offline,label="free")
plt.title('stacked bar graph')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["samsung", "apple", "others"]
online = [60, 50, 40]
offline = [75, 80, 40]
free = [50, 40, 80]

plt.bar(labels, online, label="online")
plt.bar(labels, offline, bottom=online, label="offline")

# Calculate bottom for free = online + offline (element-wise)
bottom_free = [online[i] + offline[i] for i in range(len(labels))]
plt.bar(labels, free, bottom=bottom_free, label="free")

plt.title('stacked bar graph')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

labels = ["samsung", "apple", "others"]
online = np.array([60, 50, 40])
offline = np.array([75, 80, 40])
free = np.array([50, 40, 80])

plt.bar(labels, online, label="online")
plt.bar(labels, offline, bottom=online, label="offline")
plt.bar(labels, free, bottom=online + offline, label="free")

plt.title('stacked bar graph')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
course=["ai",'web dev','data science','digital']
plt.figure(figsize=(10,20))
plt.pie(x,labels=course,startangle=45,explode=[0,0,0.5,0],autopct="%1.1f%%",colors=["green","orange","black","skyblue"])
plt.legend(title="education",loc='upper left',facecolor="blue",shadow=True,edgecolor="red",fontsize=10)
plt.title("_education_")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[2,4,5,3,7]
course=["raavi","manvir","arshpreet","tarandeep","harshit"]
plt.figure(figsize=(10,20))
plt.pie(x,labels=course,startangle=45,explode=[0.1,0,0,0,0.1],autopct="%1.1f%%",textprops={"color":"blue"},colors=["green","yellow","red","skyblue","pink"])
plt.legend(title="student_marks",loc='upper left',facecolor="blue",shadow=True,edgecolor="red",fontsize=10)
plt.title("_marks_")
plt.show()
```
